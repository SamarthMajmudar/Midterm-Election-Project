```python
import pandas as pd
import statsmodels.formula.api as smf
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
from google.colab import drive
drive.mount('/content/drive')
file_path = '/content/drive/My Drive/Midterm Project Final Data/2006_to_2024_senate_polls - Oct 30 Updated.csv'
df = pd.read_csv(file_path)
```

```python
df
```

```python
df = df[(df.Days_Before_Election >= 0) & (df.Days_Before_Election <= 200)]
```

```python
file_path = '/content/drive/My Drive/Midterm Project Final Data/2026 Midterm Polls Final.csv'
df_2026 = pd.read_csv(file_path)
```

```python
df_2026
```

```python
df_2026 = df_2026[(df_2026.Days_Before_Election <= 30)]
```

```python
df_2026
```

```python
model_without_natl_env = smf.logit("Result ~ Poll_Margin + Partisan_Lean + Dem_Incumbent + Rep_Incumbent + Midterm + Presidential_Party", data = df).fit()
```

```python
model_without_natl_env.summary()
```

```python
df_2026["Democrat's % chance of winning (model 1)"] = model_without_natl_env.predict(df_2026)
```

```python
df_2026.groupby("State").agg({"Democrat's % chance of winning (model 1)": "mean"})
```

Since we don't know what the national enviroment, one method of predicting the winners is not including the national enviroment. If we don't know the national enviroment, the model predicts a 51-49 Democratic majority. According to the modle Dems will score major flips in Maine, Texas, and Iowa, but they will be disappointed to not take states like Kansas, Ohio, and Alaska.

Now, let's add the national enviroment. Since we don't know the national enviroment, we can take a good guess from generic ballot polling data and then make a guess. To do this I will use linear regression to predict the predicted national enviroment. Currently, polling shows a D+8 national enviroment, but I expect the national enviroment to be lower for Dems since historically generic ballot polls have overestimated the Democrats.

The code below tries to approximate the national enviroment for this election cycle. This will give us an expected case senario for this year's election when it comes to the generic ballot/national enviroment.

```python
predicted = np.array([[12], [10], [-10], [0], [-3], [1], [8], [7], [-3], [0]])
actual = np.array([8, 11, -7, 1, -6, -1, 9, 3, -3, -3])
natl_env = LinearRegression()
natl_env.fit(predicted, actual)
print(natl_env.predict([[8]]))
```

The model predicts that this year, the national enviroment will be D+6.2.

```python
df_2026["Actual_National_Env"] = 6.2
```

```python
model_final = smf.logit("Result ~ Poll_Margin + Partisan_Lean + Dem_Incumbent + Rep_Incumbent + Midterm + Presidential_Party + Actual_National_Env", data = df).fit()
```

```python
model_final.summary()
```

```python
df_2026["Democrat's % chance of winning"] = model_final.predict(df_2026)
```

```python
df_2026.groupby("State").agg({"Democrat's % chance of winning": "mean"})
```

This model predicts that Dems will win the Senate 51-49.

I should note a major caveat. Historically incumbents are well liked and if an incumbent is on the ballot, their odds of winning dramatically increase. This model shows that. However, this year, many incumbents are not well liked. That leads me to make another model that does not account for incumbents.

```python
model_no_inc = smf.logit("Result ~ Poll_Margin + Partisan_Lean + Midterm + Presidential_Party + Actual_National_Env", data = df).fit()
```

```python
model_no_inc.summary()
```

```python
df_2026["Democrat's % chance of winning (without factoring in incumency)"] = model_no_inc.predict(df_2026)
```

```python
df_2026.groupby("State").agg({"Democrat's % chance of winning (without factoring in incumency)": "mean"})
```

This model gives Democrats a 52-48 seat majority in the senate.

Admittedly this model perfectly captures my predictions without the data. I don't really see how Democrats can win Iowa while at the same time lose Alaska and Ohio. So, this model makes a bit more sense, but both models are certainly valid.

Now let's create dream senarios for both Dems and Reps. This means that I will change the national enviroment. I will also make two models for each case: one with incumbency and one without.

A dream senario for Democrats would be if the polling slightly underestimates the Democratic national enviroment and the enviroment is D+9.

```python
df_2026["Actual_National_Env"] = 9
```

```python
model_dem_dream_with_inc = smf.logit("Result ~ Poll_Margin + Partisan_Lean + Dem_Incumbent + Rep_Incumbent + Midterm + Presidential_Party + Actual_National_Env", data = df).fit()
```

```python
df_2026["Democrat's dream senario with incumbency advantage"] = model_dem_dream_with_inc.predict(df_2026)
```

```python
df_2026["Dem's chance of winning in a dream year with incumbency factored in"] = model_dem_dream_with_inc.predict(df_2026)
```

```python
df_2026.groupby("State").agg({"Dem's chance of winning in a dream year with incumbency factored in": "mean"})
```

It seems like in this senario Dems win the senate 52-48 with Ohio flipping from the predicted model.

```python
model_dem_dream_without_inc = smf.logit("Result ~ Poll_Margin + Partisan_Lean + Midterm + Presidential_Party + Actual_National_Env", data = df).fit()
```

```python
df_2026["Dem's chance of winning in a dream year without incumbency factored in"] = model_dem_dream_without_inc.predict(df_2026)
```

```python
df_2026.groupby("State").agg({"Dem's chance of winning in a dream year without incumbency factored in": "mean"})
```

This would give the Dems the majority in the senate 54-46. This is the Democrats dream and best case senario for the senate.

A dream senario for the Republicans is that the polls have been off and the national enviroment is not actually D+8. And rather the national enviroment is just D+3. That would be a dream senario for Republicans because that would save the Republicans from a midterm blowout.

```python
df_2026["Actual_National_Env"] = 3
```

```python
model_rep_dream_with_inc = smf.logit("Result ~ Poll_Margin + Partisan_Lean + Dem_Incumbent + Rep_Incumbent + Midterm + Presidential_Party + Actual_National_Env", data = df).fit()
```

```python
df_2026["Dems% of winning in Republicans's dream senario with incumbency advantage"] = model_rep_dream_with_inc.predict(df_2026)
```

```python
df_2026.groupby("State").agg({"Dems% of winning in Republicans's dream senario with incumbency advantage": "mean"})
```

This model would give Republicans the senate 51-49. This model is probably the best case senario for Republicans and holding the senate 51-49 might just be the best they can do.

```python
model_rep_dream_without_inc = smf.logit("Result ~ Poll_Margin + Partisan_Lean + Midterm + Presidential_Party + Actual_National_Env", data = df).fit()
```

```python
df_2026["Dems% of winning in Republicans's dream senario without incumbency advantage"] = model_rep_dream_without_inc.predict(df_2026)
```

```python
df_2026.groupby("State").agg({"Dems% of winning in Republicans's dream senario without incumbency advantage": "mean"})
```

This model also gives Republicans a 51-49 majority in the senate.
