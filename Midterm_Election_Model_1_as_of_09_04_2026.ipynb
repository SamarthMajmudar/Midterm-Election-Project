```python
import pandas as pd
import statsmodels.formula.api as smf
```

```python
from google.colab import drive
drive.mount('/content/drive')
file_path = '/content/drive/My Drive/Midterm Election Project/Midterm Elections Model #1 Sept 4.csv'
df = pd.read_csv(file_path)
```

```python
df_train = df[df["Year"] < 2026]
df_train
```

```python
df_test = df[df["Year"] == 2026]
df_test
```

```python
log_model = smf.logit("Result ~ Poll_Margin + Partisan_Lean + President_Party + Midterm + Dem_Incumbent + Rep_Incumbent", data = df_train).fit()
log_model.summary()
```

```python
df_test["Prediction"] = log_model.predict(df_test)
```
