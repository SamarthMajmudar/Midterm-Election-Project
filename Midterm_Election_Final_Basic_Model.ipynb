```python
import pandas as pd
import statsmodels.formula.api as smf
```

```python
from google.colab import drive
drive.mount('/content/drive')
file_path = '/content/drive/My Drive/Midterm Project Final Data/Midterm Elections Poll Averages.csv'
df = pd.read_csv(file_path)
```

```python
df
```

```python
df_historical = df[(df.Year <= 2024)]
```

```python
df_historical
```

```python
df_2026 = df[df.Year == 2026]
```

```python
df_2026
```

```python
model_1 = smf.logit("Result ~ Poll_Margin + Partisan_Lean", data = df_historical).fit()
```

```python
df_2026["Dem's Chances of Winning (model1)"] = model_1.predict(df_2026)
```

```python
df_2026.groupby("State").agg({"Dem's Chances of Winning (model1)": "mean"})
```

```python
model_2 = smf.logit("Result ~ Poll_Margin + Partisan_Lean + Dem_Incumbent + Rep_Incumbent", data = df_historical).fit()
```

```python
df_2026["Dem's Chances of Winning (model2)"] = model_2.predict(df_2026)
```

```python
df_2026.groupby("State").agg({"Dem's Chances of Winning (model2)": "mean"})
```

```python
model_3 = smf.logit("Result ~ Poll_Margin + Partisan_Lean + Dem_Incumbent + Rep_Incumbent + Midterm + President_Party", data = df_historical).fit()
```

```python
df_2026["Dem's Chances of Winning (model3)"] = model_3.predict(df_2026)
```

```python
model_3.summary()
```

```python
df_2026.groupby("State").agg({"Dem's Chances of Winning (model3)": "mean"})
```
